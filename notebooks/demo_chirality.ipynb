# Chiral Coarse-Quantum Framework — Interactive Demo

This notebook demonstrates the three reference implementations:
1. Chiral spin-network simulation
2. CMB chirality estimator
3. Linguistic chirality index

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import matplotlib.pyplot as plt

from spin_network import ChiralSpinNetwork
from cmb_chirality import generate_isotropic_cmb_alm, chirality_statistic_Sl, inject_chiral_signal
from linguistic_chirality import tree_chirality_index
import networkx as nx
```

## 1. Spin-network simulation

```python
model = ChiralSpinNetwork(N=40, J=1.2, h=1e-3, kT=1.0, random_state=42)
result = model.run(n_sweeps=20_000, sample_interval=50, burn_in=2_000, show_progress=True)

plt.figure(figsize=(8, 4))
plt.plot(result.times, result.sigma_series)
plt.axhline(0, color='black', ls='--')
plt.xlabel('Sweep'); plt.ylabel(r'$\sigma$')
plt.title('Chiral order parameter evolution')
plt.show()
```

## 2. CMB chirality estimator

```python
lmax = 30
rng = np.random.default_rng(0)
alm_E, alm_B = generate_isotropic_cmb_alm(lmax, rng=rng)
_, alm_B_chiral = inject_chiral_signal(alm_E, alm_B, amplitude=0.3, l_peak=10)

ells, Sl_iso = chirality_statistic_Sl(alm_E, alm_B)
ells, Sl_chiral = chirality_statistic_Sl(alm_E, alm_B_chiral)

plt.figure(figsize=(8, 4))
plt.plot(ells, Sl_iso, label='Isotropic')
plt.plot(ells, Sl_chiral, label='Chiral signal')
plt.axhline(0, color='black', ls='--')
plt.xlabel(r'$\ell$'); plt.ylabel(r'$S_\ell$')
plt.legend(); plt.show()
```

## 3. Linguistic chirality index

```python
G = nx.DiGraph()
G.add_nodes_from([1, 2, 3])
G.add_edge(2, 1)
G.add_edge(2, 3)
print('Tree chirality tau =', tree_chirality_index(G, root=2))
```
